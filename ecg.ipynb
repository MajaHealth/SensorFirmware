```python
import numpy as np

def rms(x: np.ndarray) -> float:
    """Root-mean-square of a 1-D array (per column if 2D)."""
    return np.sqrt(np.mean(np.square(x), axis=0))

def snr_db(sig_plus_noise: np.ndarray, noise: np.ndarray) -> np.ndarray:
    """
    SNR in dB for each channel (columns).
    Uses power subtraction: P_signal = P_total − P_noise.
    Returns NaN if P_signal <=0.
    """
    p_noise  = np.square(noise).mean(axis=0)
    p_total  = np.square(sig_plus_noise).mean(axis=0)
    p_signal = np.maximum(p_total - p_noise, 1e-30)
    return 10 * np.log10(p_signal / p_noise)

def calc_vin_diff(elements,ADCMAX):
    mv = [(2 * VREF / 3.5) * (x / ADCMAX - 0.5) for x in elements]
    return mv - np.mean(mv)
    # return filtfilt(b, a, (mv - np.mean(mv)))

# ------------ Load Data ------------

fs = 533
if fs == 533:
    adc_max = ADCMAX1
else:
    adc_max = ADCMAX2

# Replace paths with your actual .npz filenames
noise120 = np.load("ecg_experiments/noise_data_120s.npz")
noise600 = np.load("ecg_experiments/noise_data_600s.npz")
data1_120 = np.load("ecg_experiments/noise_data_0.5mV_120s.npz")
data2_600 = np.load("ecg_experiments/noise_data_0.5mV_600s.npz")
data3_120 = np.load("ecg_experiments/noise_data_100mV_120s.npz")
data4_600 = np.load("ecg_experiments/noise_data_100mV_600s.npz")

# ------------ Convert to volts per lead ------------

def extract_leads(npz):
    lead_I  = calc_vin_diff(npz["lead_I"], adc_max)
    lead_II = calc_vin_diff(npz["lead_II"], adc_max)
    lead_V1 = calc_vin_diff(npz["lead_III"], adc_max)
    return np.column_stack([lead_I, lead_II, lead_V1])

noise_120_v = extract_leads(noise120)
noise_600_v = extract_leads(noise600)
d1_120_v    = extract_leads(data1_120)
d2_600_v    = extract_leads(data2_600)
d3_120_v    = extract_leads(data3_120)
d4_600_v    = extract_leads(data4_600)

# ------------ Compute SNR ------------

snr_d1_120 = snr_db(d1_120_v, noise_120_v)
snr_d2_600 = snr_db(d2_600_v, noise_600_v)
snr_d3_120 = snr_db(d3_120_v, noise_120_v)
snr_d4_600 = snr_db(d4_600_v, noise_600_v)

lead_names = ["Lead I", "Lead II", "Lead V1"]

# ------------ Print Results ------------

def print_snr(snr_array, label):
    print(f"\nSNR – {label}")
    for name, snr in zip(lead_names, snr_array):
        print(f"  {name:8s}: {snr:6.2f} dB")

print_snr(snr_d1_120, "0.5 mV, 120s")
print_snr(snr_d2_600, "0.5 mV, 600s")
print_snr(snr_d3_120, "100 mV, 120s")
print_snr(snr_d4_600, "100 mV, 600s")
```
